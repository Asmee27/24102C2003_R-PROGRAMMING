###1. Missing Data (NA)

```r
x <- NA
x

is.na(x)


x <- c(11, NA, 13, NA)
is.na(x)


x <- c(11, NA, 13, NA)
mean(x)
mean(x, na.rm = TRUE)


x <- c(11, NA, 13, NA)
which(is.na(x))


x <- c(11, NA, 13, NA)
sum(is.na(x))


x <- c(11, NA, 13, NA)
complete.cases(x)


x <- c(11, NA, 13, NA)
y <- na.omit(x)
y


x <- c(11, NA, 13, NA)
y <- na.omit(x)
mean(x)
mean(y)
```

###NA vs NULL

```r
x <- NA
x

y <- NULL
y

is.na(x)
is.null(y)
```

###4. Nested if-else-if

```r
x <- 5


if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}


x
x <- 2


if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}


x
x <- 3


if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}


x
```

###5. ifelse() Function

```r
x <- 1:10


x


ifelse(x < 6, x^2, x + 1)
x <- c(7, 9, 8, 4)


ifelse(x %% 2 == 0,
       "even number",
       "odd number")
```

###6. switch() Function
####Integer Based

```r
switch(2, "apple", "banana", "orange")
```

####String Based

```r
switch("volume",
       "colour" = "blue",
       "gender" = "male",
       "volume" = 50)
```

####No Match Cases

```r
switch("size",
       "colour" = "blue",
       "gender" = "male",
       "volume" = 50)
```

###7. which() Function

```r
x <- c(10, 15, 8, 14, 6, 12)


x


which(x == 14)
which(x != 12)
which(x > 10)
```

###8. which.min() and which.max()

```r
x <- matrix(nrow = 3,
            ncol = 3,
            data = 1:9)

which(x %% 2 == 1)


which(x %% 2 == 1,
      arr.ind = TRUE)
```

###9. which() with Matrix

```r
x <- matrix(nrow = 3,
            ncol = 3,
            data = 1:9)

which(x %% 2 == 1)
```
