```r
# ==========================================
# R PROGRAMMING - WEEK 8
# VECTOR INDEXING
# ==========================================

# 1. Create vector
x <- 1:10

x


# 2. Select values greater than 5
x[x > 5]


# 3. Select even values
x[x %% 2 == 0]


# 4. Select odd values
x[x %% 2 == 1]


# 5. Select first five values
x[1:5]


# 6. Select last five values
x[6:10]


# 7. Negative indexing
x[-(1:5)]


# 8. Same using positive indexing
x[6:10]


# ==========================================
# MISSING VALUES
# ==========================================

# 9. Create missing value
x[5] <- NA

x


# 10. Check missing values
is.na(x)


# 11. Remove missing values
y <- x[!is.na(x)]

y


# 12. Mean with missing value
mean(x)


# 13. Mean without missing value
mean(y)


# 14. Mean using na.rm
mean(x, na.rm = TRUE)


# ==========================================
# NAMED VECTOR
# ==========================================

marks <- c(
  85,
  76,
  92,
  68,
  89
)

names(marks) <- c(
  "Maths",
  "Physics",
  "Chemistry",
  "English",
  "Programming"
)

marks


# 15. Access using name
marks["Maths"]


# 16. Access multiple names
marks[
  c(
    "Maths",
    "Programming"
  )
]


# 17. Change a value
marks["English"] <- 75

marks


# ==========================================
# FORMATTING AND DISPLAY
# ==========================================

# 18. Character formatting
x <- c(
  "A",
  "BB",
  "CCC",
  "DDDD"
)

format(
  x,
  width = 7,
  justify = "centre"
)

format(
  x,
  width = 7,
  justify = "left"
)

format(
  x,
  width = 7,
  justify = "right"
)

format(
  x,
  width = 7,
  justify = "none"
)


# ==========================================
# MATRIX DISPLAY
# ==========================================

m <- matrix(
  1:6,
  nrow = 3,
  ncol = 2,
  byrow = TRUE
)

print(m)


# ==========================================
# LARGE NUMBER FORMATTING
# ==========================================

format(
  1234567,
  big.mark = ","
)

format(
  12345678,
  big.mark = ","
)

format(
  123456789,
  big.mark = " "
)
```
