```r

# ==========================================
# R PROGRAMMING - WEEK 7
# LISTS
# ==========================================

# 1. Create two matrices
x1 <- matrix(
  c(1, 2, 3, 4),
  nrow = 2,
  ncol = 2,
  byrow = TRUE
)

x2 <- matrix(
  c(5, 6, 7, 8),
  nrow = 2,
  ncol = 2,
  byrow = TRUE
)

x1
x2


# 2. Create a list containing matrices
matlist <- list(x1, x2)

matlist


# 3. Access first list element
matlist[1]


# 4. Access second list element
matlist[2]


# 5. Access actual matrix from first element
matlist[[1]]


# 6. Access actual matrix from second element
matlist[[2]]


# 7. Create list with different object types
z1 <- list(
  c("water", "juice", "lemonade"),
  rep(1:4, each = 2),
  matrix(
    data = 5:8,
    nrow = 2,
    ncol = 2,
    byrow = TRUE
  )
)

z1


# 8. Access first element
z1[[1]]


# 9. Access second element
z1[[2]]


# 10. Access third element
z1[[3]]


# 11. Access a specific value
z1[[1]][2]


# 12. Create named list
list2 <- list(
  drink1 = "water",
  drink2 = "juice",
  drink3 = "lemonade",
  drink4 = "tea",
  drink5 = "coffee",
  drink6 = "milk"
)

list2


# 13. Display names
names(list2)


# 14. Access using name
list2$drink1


# 15. Access using [[ ]]
list2[["drink2"]]


# 16. Access multiple elements
list2[1:3]


# 17. Add new element
list2$drink7 <- "smoothie"

list2


# 18. Remove element
list2$drink7 <- NULL

list2


# ==========================================
# DATE SEQUENCES
# ==========================================

# Days
seq(
  as.Date("2025-01-01"),
  by = "days",
  length = 6
)

# Months
seq(
  as.Date("2025-01-01"),
  by = "months",
  length = 6
)

# Years
seq(
  as.Date("2020-01-01"),
  by = "years",
  length = 6
)


# ==========================================
# ALPHABET SEQUENCES
# ==========================================

letters

letters[1:5]

letters[5:1]

LETTERS

LETTERS[1:5]

LETTERS[5:1]
```
