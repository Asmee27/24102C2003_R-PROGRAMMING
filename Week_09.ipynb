```r
# ------------------------------------------------------------
# 1. print() FUNCTION
# ------------------------------------------------------------

print("The zero occurs at")
print(2*pi)
print("radians")


# Multiple print statements
print("The zero occurs at")
print(2*pi)
print("radians")


# ------------------------------------------------------------
# 2. cat() FUNCTION
# ------------------------------------------------------------

cat("The zero occurs at", 2*pi, "radians.", "\n")


# ------------------------------------------------------------
# 3. DATE WITH cat()
# ------------------------------------------------------------

d = date()
d

cat("Today's date is:", d, "\n")


# ------------------------------------------------------------
# 4. sep ARGUMENT IN cat()
# ------------------------------------------------------------

x = 1:10
x

cat(x, sep = " ++ ")

cat("\n")

cat(x, sep = " / ")

cat("\n")


# ------------------------------------------------------------
# 5. USING VARIABLES WITH cat()
# ------------------------------------------------------------

x = 7

cat("The square of", x, "is", x^2, "!\n")


# ------------------------------------------------------------
# 6. FORMAT NUMERIC OUTPUT
# ------------------------------------------------------------

cat(
  "The square root of",
  x,
  "is approximately",
  format(sqrt(x), digits = 3),
  "\n"
)


# ------------------------------------------------------------
# 7. PRINTING VECTORS USING cat()
# ------------------------------------------------------------

evenno = c(2, 4, 6, 8, 10)

evenno

cat(
  "The first few even numbers are:",
  evenno,
  "...\n"
)


# ------------------------------------------------------------
# 8. fill AND labels ARGUMENTS
# ------------------------------------------------------------

x = 1:10
x

cat(
  x,
  fill = 2,
  labels = paste("(", letters[1:10], "):")
)



# ============================================================
# LECTURE 37
# PASTE FUNCTION
# ============================================================


# ------------------------------------------------------------
# 9. BASIC paste()
# ------------------------------------------------------------

paste(1:12)


# Alternative
as.character(1:12)


# ------------------------------------------------------------
# 10. JOINING STRINGS
# ------------------------------------------------------------

paste(
  "Everybody",
  "loves",
  "R Programming."
)


# ------------------------------------------------------------
# 11. paste() WITH sep
# ------------------------------------------------------------

paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "*"
)


paste(
  "Everybody",
  "loves",
  "R Programming.",
  sep = "==="
)


# ------------------------------------------------------------
# 12. paste() WITH VECTOR
# ------------------------------------------------------------

names = c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)

names


paste(
  names,
  "is",
  "a good",
  "person."
)


# ------------------------------------------------------------
# 13. collapse ARGUMENT
# ------------------------------------------------------------

names = c(
  "Prof. Singh",
  "Mr. Venkat",
  "Dr. Jha"
)


paste(
  names,
  "is",
  "a good",
  "person.",
  collapse = ", and "
)


# ------------------------------------------------------------
# 14. sep EXAMPLE
# ------------------------------------------------------------

paste(
  1,
  " is first",
  2,
  " is second",
  3,
  " is third",
  sep = "#"
)


# Observe role of spaces inside strings

paste(
  1,
  " is first ",
  2,
  " is second ",
  3,
  " is third ",
  sep = "#"
)


# ------------------------------------------------------------
# 15. CREATE VECTOR OF STRINGS
# ------------------------------------------------------------

x = paste(
  "Ex",
  1:5,
  sep = "_"
)

x


# Access individual elements

x[1]

x[2]

x[3]

x[5]


# ------------------------------------------------------------
# 16. paste() WITH collapse
# ------------------------------------------------------------

x = paste(
  "Ex",
  1:5,
  sep = "_",
  collapse = ""
)

x

x[1]


# ------------------------------------------------------------
# 17. DIFFERENCE BETWEEN VECTOR AND SINGLE STRING
# ------------------------------------------------------------

x = paste(
  "Ex",
  1:5,
  sep = "_"
)

x


x = paste(
  "Ex",
  1:5,
  sep = "_",
  collapse = ""
)

x


# ------------------------------------------------------------
# 18. paste0()
# ------------------------------------------------------------

paste0(1:10)


paste(1:10)


# ------------------------------------------------------------
# 19. paste0() VS paste()
# ------------------------------------------------------------

paste0(
  1:10,
  c(
    "st",
    "nd",
    "rd",
    rep("th", 7)
  )
)


paste(
  1:10,
  c(
    "st",
    "nd",
    "rd",
    rep("th", 7)
  )
)



# ============================================================
# LECTURE 38
# STRING SPLITTING
# strsplit()
# ============================================================


# ------------------------------------------------------------
# 20. BASIC strsplit()
# ------------------------------------------------------------

x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x


strsplit(
  x,
  split = "!"
)


# ------------------------------------------------------------
# 21. SPLIT USING &!
# ------------------------------------------------------------

x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x


strsplit(
  x,
  split = "&!"
)


# ------------------------------------------------------------
# 22. SPLIT USING !&
# ------------------------------------------------------------

x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x


y = strsplit(
  x,
  split = "!&"
)

y


# ------------------------------------------------------------
# 23. ACCESS ELEMENTS OF strsplit RESULT
# ------------------------------------------------------------

y[[1]][1]

y[[1]][2]

y[[1]][3]


# ------------------------------------------------------------
# 24. SPLITTING DATES
# ------------------------------------------------------------

dates = c(
  "2020-07-24",
  "2021-08-25",
  "2022-09-26",
  "2023-10-27"
)

dates


datesplt = strsplit(
  dates,
  "-"
)

datesplt


# ------------------------------------------------------------
# 25. CONVERT SPLIT DATES INTO CHARACTER MATRIX
# ------------------------------------------------------------

datemat = matrix(
  unlist(datesplt),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datemat


# ------------------------------------------------------------
# 26. CONVERT SPLIT DATES INTO NUMERIC MATRIX
# ------------------------------------------------------------

datematrix = matrix(
  as.numeric(unlist(datesplt)),
  nrow = 4,
  ncol = 3,
  byrow = TRUE
)

datematrix


# ------------------------------------------------------------
# 27. SPLIT STRING INTO INDIVIDUAL CHARACTERS
# ------------------------------------------------------------

strsplit(
  "Shalabh",
  split = ""
)



# ============================================================
# LECTURE 39
# MANIPULATIONS WITH STRINGS AND ALPHABETS
# ============================================================


# ------------------------------------------------------------
# 28. nchar()
# COUNT NUMBER OF CHARACTERS
# ------------------------------------------------------------

x = "R course 24.07.2022"

y = "Number of participants: 25"


x
y


nchar(x)

nchar(y)


# ------------------------------------------------------------
# 29. nzchar()
# CHECK WHETHER STRING IS NON-EMPTY
# ------------------------------------------------------------

nzchar(x)

nzchar(y)


# ------------------------------------------------------------
# 30. nzchar() WITH VECTOR
# ------------------------------------------------------------

x = c(
  "Apple",
  "Banana",
  "Cake"
)

x

nzchar(x)


# Vector containing empty string

y = c(
  "Apple",
  "",
  "Cake"
)

y

nzchar(y)


# ------------------------------------------------------------
# 31. nchar() WITH CHARACTER VECTOR
# ------------------------------------------------------------

x = c(
  "Apple",
  "Banana",
  "Cake"
)

nchar(x)


# ------------------------------------------------------------
# 32. nchar() WITH NUMBERS
# ------------------------------------------------------------

y = c(2, 4, 6)

nchar(y)


z = c(
  11,
  222,
  3333
)

nchar(z)


z1 = c(
  1.1,
  2.22,
  3.333
)

nchar(z1)


# ============================================================
# LOWERCASE AND UPPERCASE
# ============================================================


# ------------------------------------------------------------
# 33. toupper()
# ------------------------------------------------------------

x = "R course will start from 24.07.2022"

x

toupper(x)


# ------------------------------------------------------------
# 34. tolower()
# ------------------------------------------------------------

z = "INDIAN INSTITUTE OF TECHNOLOGY"

z

tolower(z)


# ============================================================
# END OF WEEK 9
# ============================================================
```
