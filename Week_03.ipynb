###Calculations with Data Vectors

```r
# Power operations
2^0.5
3^-1.5

# Power operator with scalar
 c(2,3,5,7)^2

#  Power operation with vector
 c(2,3,5,7)^c(2,3)

#  Integer division with scalar
c(2,3,5,7) %/% 2

# Integer division with data vector
 c(2,3,5,7) %/% c(2,3)

#  Modulo Division (x mod y) with scalars
7 %% 3

# Modulo Division (x mod y) with vectors
c(2,3,5,7) %% c(2,3)

#Maximum
max(1.2, 3.4, -7.8)

#Minimum
min(1.2, 3.4, -7.8)

#Arithmetic mean
mean(c(2, 3, 4))

#Abs for finding the absolute values
abs(c(-1,-2,-3,4,5))

#Sum for finding the sum of values
 sum(c(2,3,5,7))

#Log for finding the natural log (ln) of values
 log(10)
```

```r
#A matrix is a rectangular array with p rows and n columns
x = matrix(nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8))
print(x)

#Entering data column wise in a matrix
x = matrix( nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8))
print(x)

#Entering data row wise in a matrix
y = matrix( nrow=4, ncol=2, data=c(1,2,3,4,5,6,7,8), byrow = TRUE)
print(y)

#Tells the dimension of matrix
dim(x)

#Informs the dimension of matrix
attributes(x)
```
